```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

from utils import data_gen_util
import numpy as np
from utils import initvalues_util

%load_ext autoreload
%autoreload 2
```

```python
system_scale = 1
equal_mass = True
steps = 10000
train_data = True
num_bodies = 3

exp_name = ""

if equal_mass:
    exp_name = exp_name + "_equal_mass"
```

```python
tau, n, m, r, v = initvalues_util.initValues(num_bodies, equal_mass)
```

```python
ensemble_data = data_gen_util.generate_perturbed_ensemble(
    tau,
    n,
    m,
    r,
    v,
    k=10,               # 10 ensemble members, 10 for 2_3 and 5 for 10 bodies
    num_steps=steps,
    mass_perturbation_scale=0.01,
    position_perturbation_scale=0.01,
    velocity_perturbation_scale=0.01)

all_kepler_data = np.concatenate(np.array(ensemble_data['kepler_data_ensemble']), axis=0)
```

```python
# mij, r0, v0, r1, v1
if train_data:
    np.save(f"datasets/kepler_solver/simulated_train_dataset_{len(m)}_body{exp_name}_test.npy", all_kepler_data)
else:
    np.save(f"datasets/kepler_solver/simulated_dataset_{len(m)}_body{exp_name}_test.npy", all_kepler_data)
```
