```python
import numpy as np
from utils import hyperparam_search
from sklearn.metrics import mean_squared_error

%load_ext autoreload
%autoreload 2
```

```python
experiment = "2_3_body" # "10_body"
model_name = 'DT'

equal_masses = True
extra_name = ""
dataset_name = ""
```

```python
extra_name += "_2_3_body"
if equal_masses:
    extra_name += "_equal_masses"
    dataset_name = "_equal_mass"

all_kepler_data_1 = np.load(f'datasets/kepler_solver/simulated_dataset_2_body{dataset_name}.npy')
all_kepler_data_2 = np.load(f'datasets/kepler_solver/simulated_dataset_3_body{dataset_name}.npy')

all_kepler_data = np.concatenate((all_kepler_data_1, all_kepler_data_2))
```

```python
if model_name == 'RF':
    search_results_save_path = "hyperparam_search_results/results"

    hyperparams = {"input_dims": (7,),
                   "loss_func": mean_squared_error,
                    "n_estimators_options": range(1,21,2),
                    "max_depth_options": range(1,50,5),
                    "test_size": 0.2}

    hyperparam_search.do_hyperparam_search(model_name=model_name, dataset=all_kepler_data, hyperparams=hyperparams, save_path=search_results_save_path, extra_title=extra_name)
```

```python
if model_name == 'ANN':
    search_results_save_path = "hyperparam_search_results/results"

    hyperparams = {"epochs": 200,
                    "optimizer": 'adam',
                    "activation": 'leaky_relu',
                    "input_dims": (7,),
                    "output_dims": 6,
                    "loss_func": 'mean_squared_error',
                    "layer_widths": [1,3,5,10,15,20,30,40,50,75,100],
                    "num_layers": range(1,10),
                    "batch_size": 256,
                    "test_size": 0.2}

    hyperparam_search.do_hyperparam_search(model_name=model_name, dataset=all_kepler_data, hyperparams=hyperparams, save_path=search_results_save_path, extra_title=extra_name)
```
