```python
import json
from utils import hyperparam_search_plot_util as hyp_plot

%load_ext autoreload
%autoreload 2
```

```python
model_name = 'ANN'
file_path = "hyperparam_search_results/"
equal_masses = True
exp_name = "_2_3_body"
savefig = False

if equal_masses:
    exp_name += "_equal_masses"
sub_name = exp_name
```

```python
with open(file_path + 'results_' + model_name + exp_name + '.json', 'r') as file:
    hyperparam_search_results = json.load(file)
```

```python
if model_name == 'ANN':
    hyp_plot.plot_ANN_hyperparam_search(hyperparam_search_results, model_name, sub_name, savefig=savefig)
elif model_name == 'RF':
    hyp_plot.plot_RF_hyperparam_search(hyperparam_search_results, model_name, sub_name, savefig=savefig)
elif model_name == 'DT':
    hyp_plot.plot_DT_hyperparam_search(hyperparam_search_results, model_name, sub_name, savefig=savefig)
```
