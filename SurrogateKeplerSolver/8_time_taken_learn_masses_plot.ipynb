```python
import json
import matplotlib.pyplot as plt
```

```python
with open('learning_masses_results/DT_results_2_bodies.json', 'r') as f:
   DT_results_2_bodies = json.load(f)

with open('learning_masses_results/DT_results_3_bodies.json', 'r') as f:
   DT_results_3_bodies = json.load(f)
```

```python
# Create arrays for x-axis values starting from 1
x_values_2_bodies = range(1, len(DT_results_2_bodies["kepler_solver"][1]) + 1)
x_values_3_bodies = range(1, len(DT_results_3_bodies["kepler_solver"][1]) + 1)

colors = plt.rcParams['axes.prop_cycle'].by_key()['color'][:2]

# Plot using these x-values
plt.plot(x_values_2_bodies, DT_results_2_bodies["kepler_solver"][1], label="Kepler Solver (2 bodies)", color = colors[0])
plt.plot(x_values_2_bodies, DT_results_2_bodies["DT"][1], "--", label="DT-U-Final (2 bodies)", color = colors[0])
plt.plot(x_values_3_bodies, DT_results_3_bodies["kepler_solver"][1], label="Kepler Solver (3 bodies)", color = colors[1])
plt.plot(x_values_3_bodies, DT_results_3_bodies["DT"][1], "--", label="DT-U-Final (3 bodies)", color = colors[1])

# Get the maximum x value
max_x = max(len(DT_results_2_bodies["kepler_solver"][1]), 
            len(DT_results_2_bodies["DT"][1]),
            len(DT_results_3_bodies["kepler_solver"][1]), 
            len(DT_results_3_bodies["DT"][1]))

plt.xticks(range(1, max_x + 1))
plt.xlabel("Simulation Steps")
plt.ylabel("Time taken (seconds)")
plt.legend()
plt.show()
```
